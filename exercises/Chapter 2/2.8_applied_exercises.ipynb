- 8f. make new qualitative variable (Elite) by binning Top10perc variable into 2 groups based on whether or not the proportion of students coming from the top 10% of their high school classes exceeds 50%. + see how many with `value_counts()` and boxplot Outstate vs Elite

```python
college['Elite'] = pd.cut(college['Top10perc'],
                          [0,50,100],
                          labels=['No', 'Yes'])
```

```python
college.value_counts(['Elite'])
# college['Elite'].value_counts() does the same thing
# the first one uses a multi-index (so it can accomodate more than 1 column)
# the second one uses a index (so only 1 column)
# both are pandas Series
```

```python
college.boxplot(by="Elite", column="Outstate");
```
