```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ISLP import load_data
```

- 10a. Load dataset
- 10b. rows, columns, understanding
    - number of rows: 506
    - number of columns: 13

According to the website
- the rows represent the 506 suburbs in boston.
- the columns represent:
    - crim: per capita crime rate by town.
    - zn: proportion of residential land zoned for lots over 25,000 sq.ft.
    - indus: proportion of non-retail business acres per town.
    - chas: Charles River dummy variable (= 1 if tract bounds river; 0 otherwise).
    - nox: nitrogen oxides concentration (parts per 10 million).
    - rm: average number of rooms per dwelling.
    - age: proportion of owner-occupied units built prior to 1940.
    - dis: weighted mean of distances to five Boston employment centres.
    - rad: index of accessibility to radial highways.
    - tax: full-value property-tax rate per $10,000.
    - ptratio: pupil-teacher ratio by town.
    - lstat: lower status of the population (percent).
    - medv: median value of owner-occupied homes in $1000s.

- 10e. Comments on range for crime rate, tax, and pupil-teacher ratios
    - the IQR range for crime rates is 0.08 - 3.67. However, past the 95th percentile, they become very high. the maximum is 88.9
    - the IQR range for tax is 279 - 666. According to the histogram however, there is a significant number of suburbs with very high tax. specifically, there are 132 suburbs with tax=666 and 5 suburbs with tax=711.
    - the IQR range for ptratio is 17.40 - 20.2. While the ptratio varies, there is a significant number (140) of suburbs with >= 20.12 ptratio.

```python
result = boston[['crim', 'tax', 'ptratio']].agg(['min', lambda x: x.quantile(0.25), 'median', lambda x: x.quantile(0.75), 'max'])
result.index = ["Min", "25th percentile", "Median", "75th percentile", "Max"]
result
```
