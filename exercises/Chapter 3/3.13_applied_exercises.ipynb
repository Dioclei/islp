#### Simple Linear Regression with Simulated Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

rng = np.random.default_rng(1)
```

(a) / (b) / (c) Simulating data and fitting a model
- (c) Length of response is the same as the length of predictor variables, i.e. 100
- (c) True value of coefficients, $\beta_0 = -1$, $\beta_1 = 0.5$

```python
# (a) create feature X
x = rng.normal(size=100) # ~ N(0, 1)

# (b) create eps
eps = rng.normal(scale=np.sqrt(0.25), size=100) # ~ N(0, 0.25)

# (c) generate response y
y = -1 + 0.5 * x + eps

print(f"length of y is {y.shape[0]}")
```

(d) scatterplot of relationship between x and y (plotting y against x)
- Since I already know the underlying relationship is linear, before plotting, I expect to see the points follow a straight line (linear).
- The points plotted do indeed show a linear relationship.
- I initially had a misconception that I could not observe the y-intercept of -1. A y-intercept is simply the value of y at x=0, i.e. the best fit line is expected to pass through (0, -1). This is definitely plausible given the plot below.

```python
# scatterplot
fig, ax = plt.subplots(figsize=(4,4))
ax.scatter(x=x, y=y)
plt.tight_layout()
```

(e) / (f) Fitting a linear y against x model & plotting

```python
# OLS model
features = pd.DataFrame({
    'intercept': np.ones(shape=x.shape),
    'X': x,
})
result = sm.OLS(y, features).fit()
# print(result.params)
print(f"estimate of beta_0: {result.params.loc['intercept']}")
print(f"estimate of beta_1: {result.params.loc['X']}")
result.summary()
```

```python
fig, ax = plt.subplots(figsize=(4,4))
ax.scatter(x=x, y=y)

# population regression line, should be y = -1 + 0.5x (without the eps, since the E[eps] = 0)
y_pop = -1 + 0.5 * x
ax.plot(x, y_pop, c='blue', label="population regression line")

# least-squares regression line
ax.plot(x, result.fittedvalues, c='orange', label="least-squares regression line")

ax.legend()
plt.tight_layout()
```

(g) polynomial regression model y against x and x^2
- No, adding a polynomial term does not improve the fit
- R^2 didn't change at all, staying at 0.409
- p-value for xsquared is 0.959, showing that there is no evidence to reject the null hypothesis that the coefficient for xsquared is zero.

```python
features2 = pd.DataFrame({
    'intercept': np.ones(shape=x.shape),
    'x': x,
    'xsquared': np.pow(x, 2),
})
result = sm.OLS(y, features2).fit()
result.summary()
```

(h) / (i) / (j) Varying noise (i.e. increasing/decreasing eps variance) and reporting of results
- With less noise, the points are closer to the population regression line, and the model's OLS regression line follows closely.
- With more noise, the points are generally further from the population regression line, and the model's OLS regression line deviates more significantly.
- R^2 value is lower for the noisier dataset as compared to the less noisy dataset.
- As expected, the confidence intervals are smaller for the less-noise dataset and larger for the more-noise dataset.

```python
# lower / higher variance in noise
eps1 = rng.normal(scale=np.sqrt(0.1), size=100) # 0.25 - 0.15
eps2 = rng.normal(scale=np.sqrt(0.4), size=100) # 0.25 + 0.15

# same true relationship, generate response values
y1 = -1 + 0.5 * x + eps1
y2 = -1 + 0.5 * x + eps2
y_true = -1 + 0.5 * x

# fit models
result1 = sm.OLS(y1, features).fit()
result2 = sm.OLS(y2, features).fit()

# plot
fig, axs = plt.subplots(ncols=2, figsize=(10, 5))
axs[0].scatter(x, y1)
axs[0].plot(x, result1.fittedvalues, c='y', label="OLS regression line")
axs[0].plot(x, y_true, c='b', ls='--', label="Population regression line")
axs[0].legend()
axs[1].scatter(x, y2)
axs[1].plot(x, result2.fittedvalues, c='y', label="OLS regression line")
axs[1].plot(x, y_true, c='b', ls='--', label="Population regression line")
axs[1].legend()

plt.tight_layout()
```

```python
# coefficient and 95% confidence interval
pd.DataFrame(result1.params, columns=['coeff']).join(result1.conf_int().rename(columns={0: "conf_min", 1: "conf_max"}))
```

```python
# coefficient and 95% confidence interval
pd.DataFrame(result2.params, columns=['coeff']).join(result2.conf_int().rename(columns={0: "conf_min", 1: "conf_max"}))
```
