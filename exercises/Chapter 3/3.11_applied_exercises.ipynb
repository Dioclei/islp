### Investigation of t-statistic for null hypothesis $H_0:\beta=0$ in simple linear regression without an intercept

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = 2 * x + rng.normal(size=100)
```

(a) Simple linear regression of $y$ onto $x$ without intercept
- coefficient estimate $\hat{\beta}$: 1.9762
- standard error of coeff estimate: 0.117
- t-statistic, p-value: 16.898, P=0.000

comments:
- Without the intercept, the model seems to be able to estimate the coefficient well.
- The p-value is low enough that we can reject the null hypothesis that $\hat{\beta}=0$ and conclude that there is a relationship.
- The true value of the coefficient is within the 95% confidence interval.
- So why have an intercept?

```python
result = sm.OLS(y, x).fit()
result.summary()
```

(b) Simple linear regression of $x$ onto $y$ without an intercept
- coefficient estimate $\hat{\beta}$: 0.3757
- standard error of coeff estimate: 0.022
- t-statistic, p-value: 16.898, P=0.000 (same as the previous experiment)

comments:
- In this case we are estimating $x=\hat{\beta}'y$
- Because the p-value is small, we can conclude that there is a relationship.

```python
result2 = sm.OLS(x, y).fit()
result2.summary()
```

```python
fig, axs = plt.subplots(ncols=3, figsize=(12,4))
axs[0].scatter(x=x, y=y)
axs[0].plot(x, result.fittedvalues)
axs[1].scatter(x=y, y=x)
axs[1].plot(y, result2.fittedvalues)
axs[2].scatter(x, y)
axs[2].plot(x, result.fittedvalues, c='orange', label='y onto x')
axs[2].plot(result2.fittedvalues, y, c='darkblue', label='x onto y')
axs[2].set_xlabel('x')
axs[2].set_ylabel('y')
axs[2].legend()

plt.tight_layout()
```

(c) Relationship between the results in (a) and (b)
- In both (a) and (b), I am essentially drawing a line passing through (0,0) (since there is no intercept) and then minimizing the sum of squared deviations of $y$ in (a) and $x$ in (b).
- If $y=mx$, then I expected $x=m'y$ such that $m' = \frac{1}{m}$ but that is obviously not the case in the results, since I would expect the coefficient in (b) to be closer to $\frac{1}{2}$ but it's more like $0.4$-ish...
- This is because of the non-symmetric loss function, essentially in (a) we are minimizing the residuals in y, in (b) we are minimizing the residuals in x
- If we denote the equations as (a): $Y=\hat{\beta}_yX$ and (b): $X=\hat{\beta}_xY$, then
- $\hat{\beta}_y = \frac{\sum_{i=1}^ny_ix_i}{\sum_{i=1}^nx_i^2} = \frac{E[XY]}{E[X^2]}$
- $\hat{\beta}_x = \frac{\sum_{i=1}^ny_ix_i}{\sum_{i=1}^ny_i^2} = \frac{E[XY]}{E[Y^2]}$
- Then $\hat{\beta}_y = \hat{\beta}_x * \frac{\sum_{i=1}^ny_i^2}{\sum_{i=1}^nx_i^2} = \hat{\beta}_x * \frac{E[Y^2]}{E[X^2]}$
