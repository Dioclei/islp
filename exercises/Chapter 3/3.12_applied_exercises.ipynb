Simple linear regression without an intercept

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(1)
```

(a) Under what circumstance is the coefficient estimate of X onto Y the _same_ as the coefficient estimate for the regression of Y onto X?
- We know from the previous exercise that $\beta_y = \beta_x * \frac{\sum^n_{i=1}x_i^2}{\sum^n_{i=1}y_i^2}$
- For the coefficients to be the same, their ratio of squared values must be 1.
- For their ratio to be 1, it must be the case that $\sum^n_{i=1}x_i^2 = \sum^n_{i=1}y_i^2$
- This will be shown later in (c)

(b) Different coefficient estimates for X onto Y vs Y onto X
- This is already done in 3.11 but we will show it again here.

```python
x = rng.normal(size=100)
y = x + rng.normal(size=100) # in general, if there is noise in y, the sum of squares of x_i and y_i will be different.
result1 = sm.OLS(y, x).fit()
result2 = sm.OLS(x, y).fit()
print(f"y onto x: {result1.params[0]}")
print(f"x onto y: {result2.params[0]}")
```

(c) Same coefficient estimates for X onto Y vs Y onto X
- in the example below I show that a symmetric set of (x,y) + (y,x) points will have the same coefficients.
- in fact we only need the sum of squares to be the same, so it's possible to generate a set of points with the same (x, y): x=y values, and then shuffle the values around so the sum of squared values are the same, but the points appear somewhat random on a plot.

```python
x2 = np.concat((x, y)) # concatenate y coordinates to x
y2 = np.concat((y, x)) # concatenate x coordinates to y
# this essentially adds a symmetric set of flipped (y,x) points to the data
result3 = sm.OLS(y2, x2).fit()
result4 = sm.OLS(x2, y2).fit()
print(f"y2 onto x: {result3.params[0]}")
print(f"x onto y2: {result4.params[0]}")
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(10, 5))
axs[0].scatter(x, y)
axs[0].plot(x, result1.fittedvalues, c='lightblue', label="y onto x")
axs[0].plot(result2.fittedvalues, y, c='orange', label="x onto y")
axs[0].plot(np.arange(-4, 2.3, 0.1), np.arange(-4, 2.3, 0.1), c='k', ls=':', label="45 deg line")

axs[0].legend()
axs[1].scatter(x2, y2)
axs[1].plot(x2, result3.fittedvalues, c='red', label="y onto x")
axs[1].plot(result4.fittedvalues, y2, c='darkblue', label="x onto y")
axs[1].plot(np.arange(-4, 2.3, 0.1), np.arange(-4, 2.3, 0.1), c='k', ls=':', label="45 deg line")
axs[1].legend()
plt.tight_layout()
```

```python
# same x,y points, but shuffled
x3 = rng.random(size=100)
y3 = x3.copy()
rng.shuffle(y3)
result5 = sm.OLS(y3, x3).fit()
result6 = sm.OLS(x3, y3).fit()
print(f"y2 onto x: {result5.params[0]}")
print(f"x onto y2: {result6.params[0]}")
```

```python
fig, ax = plt.subplots(figsize=(6,6))
ax.scatter(x3, y3)
ax.plot(x3, result5.fittedvalues)
ax.plot(result6.fittedvalues, y3)
```
